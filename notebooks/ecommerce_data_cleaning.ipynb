```python
df.drop_duplicates(inplace=True)
```

```python
df.replace(['N/A', 'NULL', ' '], np.nan, inplace=True)
```

```python
df = df.map(lambda x: x.strip() if isinstance(x, str) else x)
```

```python
df['City'] = df['City'].str.title()
df['Customer_Name'] = df['Customer_Name'].str.title()
df['State'] = df['State'].str.title()
```

```python
df['Email'] = df['Email'].str.lower()
df = df[df['Email'].str.contains('@', na=False)]
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')
df['Delivery_Date'] = pd.to_datetime(df['Delivery_Date'], errors='coerce')
```

```python
cols = ['Qty', 'Unit_Price', 'Discount']

for x in cols:
    df[x] = pd.to_numeric(df[x], errors='coerce')
```

```python
df = df[df['Qty'] > 0]
```

```python
df['Discount'] = df['Discount'].fillna(0)

df['Phone'] = df['Phone'].fillna('Unknown')

df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
df['Sales'] = df['Qty']*df['Unit_Price']
```

```python
df
```

```python
df['Discount_Amount'] = df['Sales'] * df['Discount']/100
df['Revenue'] = df['Sales'] - df['Discount']
```

```python
df['Profit'] = df['Revenue'] * 0.20
```

```python
df
```

```python
df['Month'] = df['Order_Date'].dt.month_name()
```

```python
df
```

```python
df['Year'] = df['Order_Date'].dt.year
```

```python
df['Week_Day'] = df['Order_Date'].dt.day_name()
```

```python
df
```

```python
df.to_csv('Clean_Ecommerce.csv', index=False)
```

```python
from sqlalchemy import create_engine

engine = create_engine('postgresql+psycopg2://postgres:1234@localhost:5432/ecommerce')

df.to_sql(
    name='orders',
    con=engine,
    if_exists='replace',
    index=False
)
```

```python
df.columns = df.columns.str.lower()

df.to_sql(
    name='orders',
    con=engine,
    if_exists='replace',
    index=False
)
```
